## GDP Development Analysis (2000–2022)

From 2000 to 2022, the United States and China show the most significant GDP growth. 
China grew from about 1.2 trillion USD to about 18 trillion USD, while the US grew 
from about 10 trillion to about 25 trillion. Japan and Germany, as mature economies, 
grew more slowly. The UK shows clear dips during the 2008 financial crisis and the 
2020 pandemic. Brazil and Switzerland have smaller economies and are more affected 
by exchange rate and commodity cycles.
